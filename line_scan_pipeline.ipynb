# STXM line-scan pipeline

This notebook documents the full pipeline to load, reduce, and export NEXAFS line-scan data using the Panel-based processor. The UI has three tabs: **Setup**, **Reduction (OD)**, and **Views and export**.

**Pipeline overview**

1. **Setup**: Choose parent directory and experiment; set export parquet path; load sample config JSON (sample name to chemical formula).
2. **Reduction**: Select line-scan file; choose sample from config; set film region; define izero/sample regions (draggable bars); add/remove spot regions; compute OD per region.
3. **Views and export**: Toggle OD / mass absorption / beta; set mass-absorption fit mode; inspect reactive plot; export to parquet (OD, OD error, mass absorption, beta, and errors).

**Requirements**: `uv sync` (or `pip install -e .`) and a Jupyter kernel with the `stxm` package.
